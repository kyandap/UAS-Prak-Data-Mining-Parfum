fungsi yang mengubah komposisi menjadi daftar angka dan totalnya.

```python
import re
import pandas as pd
import numpy as np

def proses_jumlah_bibit(nilai):
    if pd.isna(nilai):
        return np.nan

    teks = str(nilai).strip().lower()

    if teks in ['', '-']:
        return np.nan

    teks = teks.replace('ml', '').replace(' ', '')

    # Titik pada 20.10 merupakan pemisah komposisi
    if teks == '20.10':
        return [20.0, 10.0]

    # Nilai 9.5 merupakan angka desimal
    if teks == '9.5':
        return [9.5]

    # Tanda + dan - merupakan pemisah komposisi
    angka = re.split(r'[+-]', teks)

    try:
        angka = [float(x) for x in angka if x != '']
        return angka if angka else np.nan
    except ValueError:
        return np.nan


def hitung_total_bibit(komposisi):
    if isinstance(komposisi, list):
        return sum(komposisi)
    return np.nan


# Uji fungsi
for nilai in ['7+8', '8 + 9 ml', '20.10', '2-6', '5-10', '9.5']:
    hasil = proses_jumlah_bibit(nilai)
    print(f'{nilai} → {hasil} → Total: {hitung_total_bibit(hasil)}')
```

#### Membuat kolom hasil preprocessing
Komposisi Bibit: menyimpan jumlah bibit setiap komponen.
Total Bibit: jumlah seluruh bibit.
Jumlah Komponen Aroma: jumlah aroma dalam satu produk.

```python
# Membuat kolom hasil preprocessing
df['Komposisi Bibit'] = df['Jumlah Bibit'].apply(proses_jumlah_bibit)

df['Total Bibit'] = df['Komposisi Bibit'].apply(hitung_total_bibit)

df['Jumlah Komponen Aroma'] = df['Komposisi Bibit'].apply(
    lambda x: len(x) if isinstance(x, list) else np.nan
)

# Melihat hasilnya
kolom_pemeriksaan = [
    'Nama Produk',
    'Jumlah Bibit',
    'Komposisi Bibit',
    'Total Bibit',
    'Jumlah Komponen Aroma'
]

display(df[kolom_pemeriksaan].head(20))
```

```python

import re

def hitung_komponen_aroma(nama):
    if pd.isna(nama):
        return np.nan

    nama = str(nama).strip()

    # Memisahkan nama berdasarkan x, +, atau &
    # yang berdiri sebagai pemisah antar nama aroma
    komponen = re.split(r'\s+(?:x|\+|&)\s+', nama, flags=re.IGNORECASE)

    return len([k for k in komponen if k.strip()])


# Menghitung ulang berdasarkan Nama Produk
df['Jumlah Komponen Aroma'] = df['Nama Produk'].apply(
    hitung_komponen_aroma
)

# Membuat kategori produk
df['Jenis Parfum'] = df['Jumlah Komponen Aroma'].apply(
    lambda x: 'Blend' if pd.notna(x) and x > 1
    else ('Single' if pd.notna(x) else np.nan)
)

# Memeriksa hasil
display(
    df[['Nama Produk', 'Jumlah Bibit', 'Total Bibit',
        'Jumlah Komponen Aroma', 'Jenis Parfum']].head(20)
)
```

membersihkan Ukuran Botol

```python

def proses_ukuran_botol(nilai):
    if pd.isna(nilai):
        return np.nan

    teks = str(nilai).strip().lower()

    # Nilai "-" ditangani terpisah menggunakan Total Bibit
    if teks == '-':
        return np.nan

    # Mengambil angka di awal teks
    cocok = re.match(r'^(\d+(?:\.\d+)?)\s*(?:ml)?', teks)

    if cocok:
        return float(cocok.group(1))

    return np.nan


# Mengonversi ukuran botol
df['Ukuran Botol (ml)'] = df['Ukuran Botol'].apply(
    proses_ukuran_botol
)

# Mengisi nilai "-" berdasarkan Total Bibit
mask_strip = (
    df['Ukuran Botol']
    .astype(str)
    .str.strip()
    .eq('-')
)

df.loc[mask_strip, 'Ukuran Botol (ml)'] = (
    df.loc[mask_strip, 'Total Bibit']
)

# Memeriksa hasil
display(
    df[['Nama Produk', 'Jumlah Bibit', 'Total Bibit',
        'Ukuran Botol', 'Ukuran Botol (ml)']].head(30)
)
```

### Normalisasi harga

```python

import re
import numpy as np
import pandas as pd

def proses_harga(nilai):
    if pd.isna(nilai):
        return np.nan

    teks = str(nilai).strip().lower()
    teks = teks.replace('rp', '').replace(' ', '')

    # Hasil akhir tertulis jelas
    if teks == '15+9=25000':
        return 25000

    # Nilai yang belum dapat dipastikan
    if teks in ['100.00', '22+27+50k', '50.000+50']:
        return np.nan

    # Membaca format harga utama + tambahan
    pola = r'^(\d{1,3}(?:[.]?\d{3})*|\d+)(?:\+(\d+)(k)?)?$'
    cocok = re.match(pola, teks)

    if cocok:
        harga_dasar = int(cocok.group(1).replace('.', ''))
        tambahan = cocok.group(2)
        satuan = cocok.group(3)

        if tambahan:
            tambahan = int(tambahan)

            # Angka tambahan tanpa k mengikuti konvensi dataset
            if satuan == 'k' or tambahan < 1000:
                tambahan *= 1000

            harga_dasar += tambahan

        return harga_dasar

    return np.nan


df['Harga (Rp)'] = df['Harga'].apply(proses_harga)

print("Jumlah missing Harga (Rp):", df['Harga (Rp)'].isna().sum())

display(
    df[['Nama Produk', 'Harga', 'Harga (Rp)']].iloc[85:100]
)
```

Menyeragamkan kapitalisasi komponen

```python
# Membuat kamus standar dari key
kamus_kapitalisasi = {
    key: variasi[0]
    for key, variasi in variasi_komponen.items()
}

# Fungsi normalisasi kapitalisasi
def normalisasi_kapitalisasi(daftar):
    if not isinstance(daftar, list):
        return daftar

    hasil = []

    for nama in daftar:
        key = re.sub(r'\s+', ' ', nama.strip()).casefold()
        nama_standar = kamus_kapitalisasi.get(key, nama.strip())
        hasil.append(nama_standar)

    return hasil

# Menerapkan normalisasi pada setiap baris
df['Komponen Aroma'] = df['Komponen Aroma'].apply(
    normalisasi_kapitalisasi
)

print("Normalisasi kapitalisasi selesai.")

display(
    df.loc[
        mask_blend,
        ['Nama Produk', 'Komponen Aroma']
    ].drop_duplicates(subset=['Nama Produk']).head(30)
)
```
