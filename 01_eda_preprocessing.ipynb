```python
import pandas as pd

url = "https://raw.githubusercontent.com/kyandap/UAS-Prak-Data-Mining-Parfum/refs/heads/main/Dataset%20-%20UAS.csv"

df = pd.read_csv(url)

df.head()
```

1. Periksa ukuran dataset

```python
# Melihat jumlah baris dan kolom
print("Ukuran dataset:", df.shape)

# Melihat nama kolom
print("\nNama kolom:")
print(df.columns.tolist())

# Melihat informasi tipe data
print("\nInformasi dataset:")
df.info()
```

Memeriksa kondisi data lebih detail

```python
# Memeriksa jumlah data kosong
print("Jumlah data kosong:")
print(df.isnull().sum())

# Melihat 10 baris pertama
print("\nContoh isi dataset:")
display(df.head(10))

# Menghitung tanda '-' pada Ukuran Botol
print("\nJumlah tanda '-' pada Ukuran Botol:")
print((df['Ukuran Botol'].astype(str).str.strip() == '-').sum())
```

2. Memeriksa data kosong

```python
# Menampilkan baris yang memiliki data kosong
data_kosong = df[df.isnull().any(axis=1)]

display(data_kosong)

# Menampilkan jumlah baris yang memiliki data kosong
print("Jumlah baris dengan data kosong:", len(data_kosong))
```

```python
# Menampilkan data kosong dengan posisi kolom yang jelas
for index, row in data_kosong.iterrows():
    print(f"\n--- Index {index} ---")
    for kolom in df.columns:
        print(f"{kolom}: {row[kolom]}")
```

Memeriksa nomor data duplikat

```python
# Membuat salinan dataset asli
df_raw = df.copy()

# Memeriksa nomor yang terduplikasi
duplikat_no = df[df.duplicated(subset=['No'], keep=False)]

print("Jumlah baris dengan No terduplikasi:",
      len(duplikat_no))

display(duplikat_no.sort_values('No'))
```

Memastikan apakah nomor berulang dalam kelompok yang sama

```python
# Memeriksa duplikasi berdasarkan Kelompok dan No
duplikat_kelompok = df[
    df.duplicated(subset=['Kelompok', 'No'], keep=False)
]

print("Jumlah baris dengan kombinasi Kelompok dan No terduplikasi:",
      len(duplikat_kelompok))

display(
    duplikat_kelompok.sort_values(['Kelompok', 'No'])
)
```

Memeriksa duplikasi data sebenarnya

```python
# Memeriksa duplikasi berdasarkan seluruh kolom
duplikat_baris = df[df.duplicated(keep=False)]

print("Jumlah baris yang terindikasi duplikat:",
      len(duplikat_baris))

display(duplikat_baris)
```
